```python
import joblib
import numpy as np
import os
import time
import gc

print(">> Starting Full 5D Data Shatter Process...")
start_t = time.time()

out_dir = "mmap_data"
os.makedirs(out_dir, exist_ok=True)

# =========================================================
# 1. SHATTER MHD DATA (14 Variables)
# =========================================================
print("\nLoading MHD Pickle into Mac Memory...")
interps = joblib.load("Interpolators_4D_Float32.pkl")

print("Saving MHD Grids...")
np.save(f"{out_dir}/mhd_grid_t.npy", interps['Bx'].grid[0].astype(np.float32))
np.save(f"{out_dir}/mhd_grid_x.npy", interps['Bx'].grid[1].astype(np.float32))
np.save(f"{out_dir}/mhd_grid_y.npy", interps['Bx'].grid[2].astype(np.float32))
np.save(f"{out_dir}/mhd_grid_z.npy", interps['Bx'].grid[3].astype(np.float32))

mhd_keys = [
    'Bx', 'By', 'Bz', 'Vx', 'Vy', 'Vz', 
    'gradBx', 'gradBy', 'gradBz', 
    'kappaX', 'kappaY', 'kappaZ', 
    'n_bnd', 'kT_bnd'
]

for k in mhd_keys:
    print(f"  -> Saving MHD {k}...")
    np.save(f"{out_dir}/mhd_{k}.npy", interps[k].values.astype(np.float32))

print("Clearing MHD from RAM...")
del interps
gc.collect()

# =========================================================
# 2. SHATTER GAP DATA (9 Variables)
# =========================================================
print("\nLoading Tsyganenko Gap Pickle into Mac Memory...")
gap_interps = joblib.load("IGRF_Baked_Float32.pkl")

# (Assuming you need the Gap grids based on previous logic, but skipping if not required)
# np.save(f"{out_dir}/gap_grid_x.npy", gap_interps['Bx'].grid[0].astype(np.float32))
# ...

gap_keys = [
    'Bx', 'By', 'Bz', 
    'gradBx', 'gradBy', 'gradBz', 
    'kappaX', 'kappaY', 'kappaZ'
]

for k in gap_keys:
    print(f"  -> Saving Gap {k}...")
    np.save(f"{out_dir}/gap_{k}.npy", gap_interps[k].values.astype(np.float32))

print("Clearing Gap data from RAM...")
del gap_interps
gc.collect()

print(f"\n>> Shatter Complete in {time.time() - start_t:.1f}s!")
print(f"Transfer the '{out_dir}' folder to your Linux server.")
```
